After loading the datasets, it’s important to check for duplicate rows, as these can lead to inaccurate analysis.

This code loops through each DataFrame (`df1`, `df2`, and `df3`) and counts the number of duplicate rows using the `duplicated()` function. The result shows how many duplicate rows exist in each dataset, which can then be addressed if necessary.

```python
# List of DataFrames
dataframes = [df1, df2, df3]

# Loop through each DataFrame to check for duplicates
for i, df in enumerate(dataframes, 1):
  num_duplicates = df.duplicated().sum()
  print(f"Number of duplicates in DataFrame {i}: {num_duplicates}")
```

The `drop_duplicates()` function removes any duplicate rows from each DataFrame. The `inplace=True` argument ensures that the changes are applied directly to the DataFrame.

Once duplicates are removed, the code checks each DataFrame again for any remaining duplicates using the `duplicated()` function. If duplicates are found, they are printed for further review; otherwise, a message is displayed confirming that no duplicates remain.

This two-step process ensures that the data is cleaned of duplicates.

```python
# List of DataFrames
dataframes = [df1, df2, df3]

# Loop to remove duplicates from each DataFrame
for df in dataframes:
  df.drop_duplicates(inplace=True)

# Loop to check for duplicates in each DataFrame and print them if found
for df in dataframes:
  duplicates = df[df.duplicated()]
  if not duplicates.empty:
    print(f"Duplicates found:\n{duplicates}")
  else:
    print("No duplicates found.")
```

This code loops through each DataFrame (`df1`, `df2`, `df3`), using the `isnull().sum()` function to calculate the number of missing values in each column. Only columns with missing values are displayed, along with the total count of missing entries for easier identification.

```python
# List of DataFrames
dataframes = [df1, df2, df3]

# Loop through each DataFrame to check for missing values
for i, df in enumerate(dataframes, 1):
    print(f"Missing Values in DataFrame {i}:")
    # Calculate the sum of missing values for each column
    missing_values = df.isnull().sum()

    # Print columns with missing values
    print(missing_values[missing_values > 0])

    print("\n" + "="*50 + "\n")  # Separator for readability
```

Understanding the data types of each column in a dataset is crucial for performing accurate data analysis. The `dtypes` attribute in pandas provides the data type of each column in a DataFrame. This helps verify that the data types are appropriate for the operations intended, such as ensuring numerical columns are treated as floats and date columns are recognized as datetime objects.

The following code loops through each DataFrame (`df1`, `df2`, `df3`) and prints the data types for each column. A separator is added for readability between the outputs of different DataFrames.

```python
# List of DataFrames
dataframes = [df1, df2, df3]

# Loop through each DataFrame to check data types
for i, df in enumerate(dataframes, 1):
    print(f"Data Types in DataFrame {i}:")

    # Print data types of each column
    print(df.dtypes)

    print("\n" + "="*50 + "\n")  # Separator for readability
```

Pandas has recognized all columns as object, a generic type for strings or mixed data. However, each column should be converted to the correct data type for improved performance and clarity. Applying these changes will optimize the data and make it more efficient for analysis.

```python
# List of DataFrames to check and their respective date columns
dfs = [
    (df1, 'SignupDate'),
    (df2, 'TransactionDate'),
    (df3, 'InteractionDate')
]

# Loop over each DataFrame and its respective date column
for i, (df, date_column) in enumerate(dfs, 1):
    # Attempt to convert the date column to datetime, keeping original values intact
    converted_dates = pd.to_datetime(df[date_column], errors='coerce')

    # Find rows where the conversion failed (NaT values) and get the original date formats
    invalid_dates = df.loc[converted_dates.isna(), date_column]

    # Count the number of invalid date formats
    count_invalid = invalid_dates.shape[0]

    # Print the count and the original invalid date formats
    print(f"DataFrame {i} - Column '{date_column}':")
    print(f"Number of invalid date formats: {count_invalid}")

    if count_invalid > 0:
        print("Invalid date formats (original values):")
        print(invalid_dates.values)  # Print the original invalid date values
    else:
        print("No invalid date formats found.")

    print("="*50)  # Separator for better readability
```

Converting 'SignupDate' to DateTime

This code is used to convert a 'SignupDate' column in a pandas DataFrame to a proper `datetime` format while handling different formats and invalid dates.

1. Identify Specific Date Format (`dd/mm/yyyy`):
   - The code first checks which rows follow the `dd/mm/yyyy` format using a regular expression.
   
2. Convert to DateTime:
   - For rows that match this format, it converts the dates into proper `datetime` format. If the format doesn't match, it sets the value to `NaT` (Not a Time).

3. Handle Remaining Dates:
   - The rest of the 'SignupDate' column is also converted to `datetime`, replacing any invalid dates with `NaT`.

4. Check for Invalid Dates:
   - The code checks how many dates could not be converted (i.e., remain as `NaT`) and prints the count.

5. Check Data Type:
   - It prints the data type of 'SignupDate' to verify that it has been converted to `datetime`.

This ensures that all date formats are handled properly, and any invalid dates are tracked.


```python
# Convert 'SignupDate' to datetime, handling different formats
invalid_format = df1['SignupDate'].str.contains(r'\d{2}/\d{2}/\d{4}', na=False)
df1.loc[invalid_format, 'SignupDate'] = pd.to_datetime(df1.loc[invalid_format, 'SignupDate'], format='%d/%m/%Y', errors='coerce')

df1['SignupDate'] = pd.to_datetime(df1['SignupDate'], errors='coerce')

# Check for invalid dates
invalid_dates = df1['SignupDate'][df1['SignupDate'].isnull()]
print(f"Number of invalid dates: {len(invalid_dates)}")

# Print data type of 'SignupDate'
print(df1['SignupDate'].dtype)
```

Changing Data Types in a DataFrame

This code snippet focuses on changing the data types of several columns in a pandas DataFrame (`df1`). Here's a breakdown of each step:

1. Changing Data Types of Specific Columns:
   - The `CustomerID` column is converted to a `string` data type.
   - The `Gender` and `IncomeLevel` columns are converted to `category` type. Categories are useful for columns with a limited set of possible values, optimizing memory usage.
   - The `SignupDate` column is converted to `datetime` format, specifying the correct date format (`dd/mm/yyyy`).

2. Handling Numeric Data with 'Age':
   - The `Age` column is converted to a numeric type, with `errors='coerce'` ensuring that any invalid values are replaced with `NaN` (Not a Number), which represents missing or invalid data.

3. Verifying the Changes:
   - Finally, the data types of all columns in the DataFrame are printed to confirm that the changes were applied correctly.

This ensures that the DataFrame uses the appropriate data types for each column, which can improve both memory efficiency and the correctness of operations on the data.


```python
# Changing data types
df1['CustomerID'] = df1['CustomerID'].astype('string')  # Change to string
df1['Gender'] = df1['Gender'].astype('category')  # Change to category
df1['IncomeLevel'] = df1['IncomeLevel'].astype('category')  # Change to category
df1['SignupDate'] = pd.to_datetime(df1['SignupDate'], format='%d/%m/%Y')  # Change to datetime, specifying the correct format

# Convert 'Age' to numeric, replacing invalid values with NaN
df1['Age'] = pd.to_numeric(df1['Age'], errors='coerce')
# NaN (Not a Number) is a special floating-point value that represents missing or invalid data.

# Verify the changes
print(df1.dtypes)
```
